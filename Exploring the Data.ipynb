```python
import pandas as pd 

df = pd.read_csv("Portland_Crime_Data.csv")
```

```python
df.drop(columns=["Case Number", "Number of Records", "Occur Date", "Offense Count", "OpenDataX", "OpenDataY", "ReportMonthYear"], inplace=True)
```

```python
df.rename(columns={"Crime Against": "Crime Type", "Occur Time": "Occurrence Time", "OpenDataLat": "Latitude", "OpenDataLon": "Longitude"}, inplace=True)
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
import matplotlib.pyplot as plt
plt.style.use('ggplot') 
```

```python
df["crime_type"].value_counts().plot(kind="bar", figsize=(12, 6), title="Crime Type Distribution")
plt.xlabel("Crime Type")
plt.show()
```

```python
df["offense_type"].value_counts().plot(kind="bar", figsize=(12, 6), title="Offense Type Distribution")
plt.xlabel("Offense Type")
plt.show()
```

```python
df["hour"] = df["occurrence_time"].astype(str).str.zfill(4).str[:2].astype(int)
```

```python
df["hour"].value_counts().sort_index().plot(kind="bar", figsize=(12,6), title="Crimes by Hour of Day")
plt.xlabel("Hour")
plt.ylabel("Number of Crimes")
plt.show()
```

```python
df["neighborhood"].value_counts().head(15).plot(kind="bar", figsize=(12,6), title="Top 15 Neighborhoods by Crime Count")
plt.xlabel("Neighborhood")
plt.show()
```

```python
df.to_csv("Portland_Crime_Data_Cleaned.csv", index=False)
```
